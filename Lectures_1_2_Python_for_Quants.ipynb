##From Problem to Result through Programming

Programming is a way of **solving problem**s with the help of a computer.

A problem usually starts with a question or a task.

To solve the problem we first need a **plan**.

This plan is called an **algorithm**.

An algorithm describes the **sequence of steps** that must be performed.



The algorithm is then translated into a **programming language**.

The resulting program can be **executed by a computer**.

The execution of the program produces a **result**.



The process can be summarized as follows:

**Problem → Algorithm → Program → Result**

## Simple return

A simple return measures the relative change between two prices.

The formula is the final price divided by the initial price minus one.

This is one of the most common calculations in finance.

```python
initial_price = 100
final_price = 108

simple_return = final_price / initial_price - 1

print(f"Simple return: {simple_return:.2%}")
```

## Profit and loss

Profit and loss measures the absolute gain or loss on a position.

It depends on the number of units and the change in price.

This example uses only basic arithmetic.

```python
shares = 50
buy_price = 100
sell_price = 108

profit_loss = shares * (sell_price - buy_price)

print(f"Profit or loss: {profit_loss:.2f}")
```

## Comparisons

## Common comparison operators

The operator `>` means greater than.

The operator `<` means less than.

The operator `>=` means greater than or equal to.

The operator `<=` means less than or equal to.

## Classifying a return

## Logical operators

## A simple investment rule

A simple rule can combine expected return and risk.

This example accepts an investment when the expected return is high enough and the risk is not too high.

The thresholds are part of the model assumptions.

```python
expected_return = 0.07
volatility = 0.18

if expected_return >= 0.06 and volatility <= 0.20:
    decision = "accept"
else:
    decision = "reject"

print(f"Decision: {decision}")
```

## Buy sell or hold

A trading rule maps data into an action.

This example uses one signal value.

A positive signal suggests buying and a negative signal suggests selling.

```python
signal = 0.35

if signal > 0.20:
    action = "buy"
elif signal < -0.20:
    action = "sell"
else:
    action = "hold"

print(f"Trading action: {action}")
```

## Stop loss rule

A stop loss rule limits losses.

The program checks whether the current loss exceeds a chosen threshold.

This is a simple example of risk control.

```python
entry_price = 100
current_price = 92
stop_loss = -0.05

current_return = current_price / entry_price - 1

if current_return <= stop_loss:
    print("Stop loss triggered")
else:
    print("Position remains open")

print(f"Current return: {current_return:.2%}")
```

A program can check whether a position is within the limit.

This type of logic is simple but important.

```python
position_value = 1_250_000
risk_limit = 1_000_000

if position_value > risk_limit:
    print("Risk limit exceeded")
else:
    print("Risk limit respected")
```

## Nested decisions

Sometimes a program needs one decision inside another decision.

Nested conditions should be used with care.

Too many levels can make code hard to read.

```python
expected_return = 0.09
volatility = 0.22
liquidity_score = 0.80

if expected_return > 0.06:
    if volatility < 0.25:
        if liquidity_score > 0.70:
            print("Candidate investment")
        else:
            print("Liquidity is too low")
    else:
        print("Volatility is too high")
else:
    print("Expected return is too low")
```

## A clearer version with combined conditions

The same idea can often be written more clearly.

Combining simple conditions can reduce nesting.

This improves readability.

```python
expected_return = 0.09
volatility = 0.22
liquidity_score = 0.80

candidate = (
    expected_return > 0.06
    and volatility < 0.25
    and liquidity_score > 0.70
)

if candidate:
    print("Candidate investment")
else:
    print("Investment does not satisfy all criteria")
```

## Guarding against division by zero

A program should check risky operations before performing them.

Division by zero is a common example.

The condition prevents an invalid calculation.

```python
initial_price = 0
final_price = 105

if initial_price == 0:
    print("Return cannot be computed because the initial price is zero.")
else:
    simple_return = final_price / initial_price - 1
    print(f"Simple return: {simple_return:.2%}")
```

## Integrated example

The next example combines arithmetic and conditional logic.

It computes a return and then applies a stop loss rule.

This pattern appears frequently in quantitative applications.

```python
entry_price = 50
current_price = 47
position_size = 200
stop_loss_threshold = -0.04

current_return = current_price / entry_price - 1
profit_loss = position_size * (current_price - entry_price)

if current_return <= stop_loss_threshold:
    action = "close position"
else:
    action = "keep position"

print(f"Current return: {current_return:.2%}")
print(f"Profit or loss: {profit_loss:.2f}")
print(f"Action: {action}")
```

## Integrated example with user defined assumptions

The values at the top of the cell are assumptions.

Changing assumptions changes the result.

This is a simple form of scenario analysis.

```python
initial_capital = 100_000
expected_return = 0.07
maximum_acceptable_risk = 0.18
estimated_risk = 0.16

expected_profit = initial_capital * expected_return

if expected_return > 0.05 and estimated_risk <= maximum_acceptable_risk:
    recommendation = "consider investment"
else:
    recommendation = "do not invest"

print(f"Expected profit: {expected_profit:.2f}")
print(f"Estimated risk: {estimated_risk:.2%}")
print(f"Recommendation: {recommendation}")
```

## Common beginner mistakes

Using `=` instead of `==` in a condition is a common mistake.

Forgetting indentation is another common mistake.

Comparing percentages written in different forms can also lead to errors.

The value 5 means five units while the value 0.05 means five percent.

```python
required_return = 0.05
observed_return = 5 / 100

print(required_return == observed_return)
```

## Mini case study

A small fund evaluates an asset.

The fund accepts the asset only if the expected return is at least 6 percent.

The volatility must be no more than 20 percent.

The liquidity score must be at least 70 percent.

The asset must also have a positive expected profit for the proposed investment size.

```python
investment_size = 250_000
expected_return = 0.075
volatility = 0.19
liquidity_score = 0.74

expected_profit = investment_size * expected_return

meets_return_requirement = expected_return >= 0.06
meets_risk_requirement = volatility <= 0.20
meets_liquidity_requirement = liquidity_score >= 0.70
has_positive_profit = expected_profit > 0

if (
    meets_return_requirement
    and meets_risk_requirement
    and meets_liquidity_requirement
    and has_positive_profit
):
    decision = "approved"
else:
    decision = "not approved"

print(f"Expected profit: {expected_profit:.2f}")
print(f"Decision: {decision}")
```
