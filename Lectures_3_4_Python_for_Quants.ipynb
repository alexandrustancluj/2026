# Repetitive Computation and Iterative Algorithms

Many problems require the same action to be performed many times.

A computer is useful when a task is repetitive.

A loop allows a program to repeat instructions automatically.

Loops help us avoid writing the same code again and again.


## Problem → Algorithm → Program → Result

A repeated task can be described as an algorithm.

The algorithm specifies what must be repeated.

The program implements this repetition through a loop.

The result is obtained after the loop has finished.


```python
print("Trading day 1")
print("Trading day 2")
print("Trading day 3")
print("Trading day 4")
print("Trading day 5")
```

The previous program works.

It is not flexible.

If we need one hundred days then we do not want to write one hundred print statements.

A loop solves this problem.


```python
for day in range(1, 6):
    print(f"Trading day {day}")
```

## The for loop

A `for` loop repeats a block of code.

The loop variable takes one value at a time.

The indented block is executed for each value.

This structure is useful when the number of repetitions is known.


```python
for number in range(5):
    print(number)
```

## The range function

The `range()` function creates a sequence of integer values.

The endpoint is not included.

The expression `range(1, 6)` produces the values 1 to 5.


```python
for number in range(1, 6):
    print(number)
```

## Custom steps

The `range()` function can also use a step.

The third argument controls the distance between values.

This is useful when we need every second value or every fifth value.


```python
for number in range(0, 21, 5):
    print(number)
```

## Iterating over text

A loop can iterate over a string.

Each character is processed one at a time.

This shows that Python loops are not limited to numbers.


```python
ticker = "AAPL"

for character in ticker:
    print(character)
```

## Iterating over a list

A list stores multiple values.

A `for` loop can process each element from a list.

This pattern is common in data analysis.


```python
prices = [100, 101, 103, 102, 105]

for price in prices:
    print(price)
```

## Accumulators

An accumulator is a variable that collects a result step by step.

It usually starts with an initial value.

The loop updates the accumulator during each iteration.


```python
values = [10, 20, 30, 40]

total = 0

for value in values:
    total = total + value

print(total)
```

## Computing an average

The average is the total divided by the number of values.

A loop can compute the total.

The length of the list gives the number of values.


```python
returns = [0.01, -0.02, 0.03, 0.015, -0.005]

total_return = 0

for value in returns:
    total_return += value

average_return = total_return / len(returns)

print(f"Average return: {average_return:.2%}")
```

## Counting with a loop

A loop can count how many times a condition is true.

This is useful when we want to summarize data.

The counter increases only when the condition is satisfied.


```python
returns = [0.01, -0.02, 0.03, 0.015, -0.005]

positive_days = 0

for value in returns:
    if value > 0:
        positive_days += 1

print(f"Number of positive days: {positive_days}")
```

## Finding the maximum value

A loop can search for the largest value.

We start with the first element.

Then each new value is compared with the current maximum.


```python
prices = [100, 101, 103, 102, 105, 99]

maximum_price = prices[0]

for price in prices:
    if price > maximum_price:
        maximum_price = price

print(f"Maximum price: {maximum_price}")
```

## Finding the minimum value

The same logic can find the smallest value.

Only the comparison changes.

This example shows a reusable algorithmic pattern.


```python
prices = [100, 101, 103, 102, 105, 99]

minimum_price = prices[0]

for price in prices:
    if price < minimum_price:
        minimum_price = price

print(f"Minimum price: {minimum_price}")
```

## Computing simple returns from prices

A price series can be transformed into a return series.

Each return uses two consecutive prices.

This is a common example of repeated computation.


```python
prices = [100, 102, 101, 105, 107]

returns = []

for index in range(1, len(prices)):
    simple_return = prices[index] / prices[index - 1] - 1
    returns.append(simple_return)

print(returns)
```

## Printing returns clearly

Raw decimal values are not always easy to read.

Formatted output makes results easier to interpret.

This is important when results are shown to other people.


```python
prices = [100, 102, 101, 105, 107]

for index in range(1, len(prices)):
    simple_return = prices[index] / prices[index - 1] - 1
    print(f"Return from day {index} to day {index + 1}: {simple_return:.2%}")
```

## Compound growth with repeated multiplication

Many growth processes are repeated over time.

The value at the end of each period becomes the starting value for the next period.

A loop represents this process naturally.


```python
capital = 10_000
annual_return = 0.05
years = 5

for year in range(1, years + 1):
    capital = capital * (1 + annual_return)
    print(f"Year {year}: {capital:.2f}")
```

## Scenario with yearly contributions

A model can include a yearly contribution.

The order of operations must be explicit.

Here the return is applied first and the contribution is added after that.


```python
capital = 10_000
annual_return = 0.05
yearly_contribution = 1_000
years = 5

for year in range(1, years + 1):
    capital = capital * (1 + annual_return)
    capital = capital + yearly_contribution
    print(f"Year {year}: {capital:.2f}")
```

## The while loop

A `while` loop repeats while a condition is true.

It is useful when the number of repetitions is not known in advance.

The condition must eventually become false.


```python
capital = 10_000
target = 15_000
annual_return = 0.05
years = 0

while capital < target:
    capital = capital * (1 + annual_return)
    years += 1

print(f"Years needed: {years}")
print(f"Final capital: {capital:.2f}")
```

## Avoiding infinite loops

A `while` loop can run forever if the condition never becomes false.

This is called an infinite loop.

The program must update the variables used in the condition.


```python
counter = 1

while counter <= 5:
    print(counter)
    counter += 1
```

## Break

The keyword `break` stops a loop immediately.

It is useful when the program has found what it needs.

It can also be used when a safety condition is reached.


```python
prices = [100, 98, 95, 89, 92]
stop_price = 90

for price in prices:
    print(f"Checking price {price}")
    if price < stop_price:
        print("Stop condition reached")
        break
```

## Continue

The keyword `continue` skips the rest of the current iteration.

The loop then moves to the next value.

It is useful when some observations should be ignored.


```python
returns = [0.01, None, 0.02, -0.01, None, 0.03]

valid_total = 0
valid_count = 0

for value in returns:
    if value is None:
        continue
    valid_total += value
    valid_count += 1

average_return = valid_total / valid_count

print(f"Average valid return: {average_return:.2%}")
```

## Mini case study for

A small investment account starts with an initial balance.

Each year the account grows by a fixed rate.

A contribution is added at the end of each year.

The program stops when the target amount is reached.


```python
balance = 5_000
target_balance = 20_000
annual_return = 0.06
annual_contribution = 1_500
year = 0

while balance < target_balance:
    year += 1
    balance *= 1 + annual_return
    balance += annual_contribution
    print(f"Year {year}: {balance:.2f}")

print(f"Target reached after {year} years.")
```

# Functions and Modular Programming

A function is a named block of code.

A function performs a specific task.

Functions help us avoid repetition.

They also help us organize programs into smaller parts.


## Why functions matter

Without functions we often copy the same logic many times.

Copied code is difficult to maintain.

A function allows us to write the logic once and reuse it when needed.


```python
initial_price = 100
final_price = 105
return_1 = final_price / initial_price - 1

initial_price = 200
final_price = 210
return_2 = final_price / initial_price - 1

print(f"First return: {return_1:.2%}")
print(f"Second return: {return_2:.2%}")
```

## Defining a function

A function is defined with the keyword `def`.

The function name should describe the task.

The body of the function is indented.


```python
def say_hello():
    print("Hello")

say_hello()
```

## Parameters

A parameter is an input to a function.

Parameters make a function flexible.

The same function can be used with different values.


```python
def greet(name):
    print(f"Hello {name}")

greet("Ana")
greet("Mihai")
```

## Return values

A function can return a result.

The keyword `return` sends the result back to the caller.

Returned values can be stored in variables.


```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 15)

print(result)
```

## A function for simple return

This function implements the simple return formula.

The function receives the initial price and the final price.

It returns the computed return.


```python
def simple_return(initial_price, final_price):
    return final_price / initial_price - 1

result = simple_return(100, 110)

print(f"Return: {result:.2%}")
```

## Reusing the function

A function becomes more useful when it is called many times.

The logic is written once.

The inputs change from one call to another.


```python
return_a = simple_return(100, 105)
return_b = simple_return(200, 190)
return_c = simple_return(50, 55)

print(f"Return A: {return_a:.2%}")
print(f"Return B: {return_b:.2%}")
print(f"Return C: {return_c:.2%}")
```

## Functions and validation

A function can check whether the input values make sense.

This makes the program safer.

The next function avoids division by zero.


```python
def safe_simple_return(initial_price, final_price):
    if initial_price == 0:
        return None
    return final_price / initial_price - 1

result = safe_simple_return(0, 110)

print(result)
```

## Clear output with a function

A calculation function should usually return a value.

A reporting function may print a message.

Keeping these roles separate often makes code easier to reuse.


```python
def format_percentage(value):
    return f"{value:.2%}"

result = simple_return(100, 108)

print(format_percentage(result))
```

## Local variables

Variables created inside a function are local.

They exist only inside that function.

This helps prevent accidental changes in other parts of the program.


```python
def compute_profit(shares, buy_price, sell_price):
    price_change = sell_price - buy_price
    profit = shares * price_change
    return profit

profit_value = compute_profit(100, 50, 53)

print(profit_value)
```

## Function names

Good function names usually contain a verb.

Examples include compute return calculate average and validate input.

A good name makes the program easier to read.


```python
def compute_average(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)

returns = [0.01, -0.02, 0.03, 0.015]

average = compute_average(returns)

print(f"Average: {average:.2%}")
```

## Functions that use loops

A function can contain a loop.

This allows us to package an algorithm inside a reusable component.

The caller does not need to know all internal steps.


```python
def count_positive_values(values):
    count = 0
    for value in values:
        if value > 0:
            count += 1
    return count

returns = [0.01, -0.02, 0.03, 0.015, -0.005]

positive_count = count_positive_values(returns)

print(positive_count)
```

## Computing returns from a price list

A function can transform one list into another list.

Here the input is a list of prices.

The output is a list of simple returns.


```python
def compute_returns(prices):
    returns = []
    for index in range(1, len(prices)):
        value = prices[index] / prices[index - 1] - 1
        returns.append(value)
    return returns

prices = [100, 102, 101, 105, 107]

returns = compute_returns(prices)

print(returns)
```

## A function for maximum drawdown

Drawdown measures the decline from a previous high.

It is a useful example because it combines loops conditions and variables.

The implementation here is simplified for teaching.


```python
def compute_max_drawdown(prices):
    peak = prices[0]
    max_drawdown = 0

    for price in prices:
        if price > peak:
            peak = price

        drawdown = price / peak - 1

        if drawdown < max_drawdown:
            max_drawdown = drawdown

    return max_drawdown

prices = [100, 110, 105, 120, 90, 95]

max_drawdown = compute_max_drawdown(prices)

print(f"Maximum drawdown: {max_drawdown:.2%}")
```

## Default parameter values

A function can define default values for some parameters.

The default is used when the caller does not provide a value.

This can make functions easier to use.


```python
def future_value(principal, rate, years=1):
    return principal * (1 + rate) ** years

print(future_value(1000, 0.05))
print(future_value(1000, 0.05, 5))
```

## Keyword arguments

Arguments can be passed by name.

This often improves readability.

It is especially useful when a function has several inputs.


```python
result = future_value(principal=10_000, rate=0.06, years=3)

print(f"{result:.2f}")
```

## A function with a decision rule

Functions can return decisions.

The next function accepts or rejects an investment based on two conditions.

The thresholds are explicit in the function definition.


```python
def evaluate_investment(expected_return, volatility):
    if expected_return >= 0.06 and volatility <= 0.20:
        return "accept"
    return "reject"

decision = evaluate_investment(0.07, 0.18)

print(decision)
```

## Small functions are often better

A function should usually do one clear thing.

Large functions are harder to test.

Small functions can be combined into larger programs.


```python
def is_positive(value):
    return value > 0

def classify_return(value):
    if is_positive(value):
        return "positive"
    if value < 0:
        return "negative"
    return "zero"

print(classify_return(0.03))
print(classify_return(-0.01))
print(classify_return(0.00))
```

## Exercise 6

Write a function that computes the average of a list.

The function should return `None` if the list is empty.

This prevents division by zero.


```python
def average(values):
    if len(values) == 0:
        return None

    total = 0
    for value in values:
        total += value

    return total / len(values)

print(average([10, 20, 30]))
print(average([]))
```

## Integrated example

The next program uses several small functions.

Each function has one responsibility.

Together they form a simple analysis workflow.


```python
def compute_returns(prices):
    returns = []
    for index in range(1, len(prices)):
        value = prices[index] / prices[index - 1] - 1
        returns.append(value)
    return returns

def average(values):
    if len(values) == 0:
        return None

    total = 0
    for value in values:
        total += value

    return total / len(values)

def count_negative_values(values):
    count = 0
    for value in values:
        if value < 0:
            count += 1
    return count

prices = [100, 101, 99, 103, 104, 102]

returns = compute_returns(prices)
average_return = average(returns)
negative_days = count_negative_values(returns)

print(f"Returns: {returns}")
print(f"Average return: {average_return:.2%}")
print(f"Negative days: {negative_days}")
```

##  Reproducibility

A function makes a calculation reproducible.

The same input should produce the same output.

This is important in scientific and quantitative work.

A calculation that cannot be reproduced is difficult to trust.


```python
def reproducible_calculation(x):
    return x * x + 2 * x + 1

print(reproducible_calculation(5))
print(reproducible_calculation(5))
```
