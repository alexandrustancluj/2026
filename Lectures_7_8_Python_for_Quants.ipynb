# Exception Handling and Robust Software Development

A program may fail for many reasons.

The code may contain a syntax error.

The input data may be invalid.

A file may be missing.

A calculation may not be possible.

Robust programs anticipate problems and handle them clearly.


## Why errors matter

An error is not only a technical problem.

An error can produce wrong results.

Wrong results can lead to wrong decisions.

In quantitative work this can be very costly.


```python
price_initial = 100
price_final = 110

simple_return = price_final / price_initial - 1

print(f"Return: {simple_return:.2%}")
```

## Runtime errors

A runtime error appears while the program is running.

The code is syntactically correct.

The problem appears because the operation cannot be completed.


```python
price_initial = 0
price_final = 110

# This line would cause a ZeroDivisionError.
# simple_return = price_final / price_initial - 1

print("The formula cannot be applied when the initial price is zero.")
```

## Exceptions

An exception is an event that interrupts normal program execution.

Python raises an exception when something goes wrong.

The programmer can decide how to handle the exception.


```python
text_value = "100"

number_value = int(text_value)

print(number_value)
```

## The try except structure

The code inside `try` is attempted first.

If no exception appears then the program continues normally.

If the specified exception appears then the `except` block is executed.


```python
text_value = "one hundred"

try:
    number_value = int(text_value)
    print(number_value)
except ValueError:
    print("The value cannot be converted to an integer.")
```

## Division by zero

Some formulas are not always valid.

A return cannot be computed when the initial price is zero.

The exception can be handled without stopping the notebook.


```python
initial_price = 0
final_price = 110

try:
    simple_return = final_price / initial_price - 1
    print(f"Return: {simple_return:.2%}")
except ZeroDivisionError:
    print("Return cannot be computed because the initial price is zero.")
```

## Handling invalid input

Input values may come from users or external files.

A program should not assume that all inputs are valid.

Validation is an important part of robust software.


```python
raw_value = "105.50"

try:
    price = float(raw_value)
    print(f"Valid price: {price}")
except ValueError:
    print("The input is not a valid price.")
```

## Handling several exception types

Different problems may require different messages.

A program can handle several exception types separately.

This helps the user understand what went wrong.


```python
prices = [100, 102, 101]
index = 5

try:
    selected_price = prices[index]
    result = 1000 / selected_price
    print(result)
except IndexError:
    print("The selected index does not exist.")
except ZeroDivisionError:
    print("The selected price cannot be zero.")
```

## Avoiding broad exception handling

Catching every possible exception is usually not a good habit.

It may hide real problems.

It is better to catch the specific exceptions that we expect.


```python
raw_price = "not available"

try:
    price = float(raw_price)
except ValueError:
    price = None

print(price)
```

## The else block

The `else` block runs only when no exception occurs.

It can be used for code that depends on a successful operation.

This keeps the logic clear.


```python
raw_price = "101.25"

try:
    price = float(raw_price)
except ValueError:
    print("Invalid price")
else:
    print(f"The price was converted successfully: {price}")
```

## The finally block

The `finally` block runs in all cases.

It runs whether an exception occurs or not.

It is often used for cleanup tasks.


```python
raw_price = "101.25"

try:
    price = float(raw_price)
    print(f"Price: {price}")
except ValueError:
    print("Invalid price")
finally:
    print("Validation attempt finished.")
```

## Raising exceptions

A function can raise an exception deliberately.

This is useful when invalid input should stop the calculation.

The error message should be clear and useful.


```python
def compute_return(initial_price, final_price):
    if initial_price <= 0:
        raise ValueError("Initial price must be positive.")
    return final_price / initial_price - 1

try:
    result = compute_return(0, 110)
    print(f"Return: {result:.2%}")
except ValueError as error:
    print(error)
```

## Defensive programming

Defensive programming means writing code that checks assumptions.

The program should verify inputs before using them.

This reduces the chance of silent errors.


```python
def average(values):
    if len(values) == 0:
        raise ValueError("The list of values cannot be empty.")
    return sum(values) / len(values)

try:
    print(average([]))
except ValueError as error:
    print(error)
```

## Missing values

Real datasets often contain missing values.

A robust program should decide how to treat them.

One simple approach is to skip invalid observations.


```python
raw_prices = ["100", "102", "", "105", "error", "107"]

prices = []

for raw_price in raw_prices:
    try:
        price = float(raw_price)
    except ValueError:
        continue
    prices.append(price)

print(prices)
```

## Logging invalid values

Skipping invalid values may be acceptable.

Still we may want to remember what was skipped.

This creates a simple audit trail.


```python
raw_prices = ["100", "102", "", "105", "error", "107"]

prices = []
invalid_values = []

for raw_price in raw_prices:
    try:
        price = float(raw_price)
    except ValueError:
        invalid_values.append(raw_price)
    else:
        prices.append(price)

print("Valid prices:", prices)
print("Invalid values:", invalid_values)
```

## Exercise 1

Write a function called `safe_divide`.

The function receives two numbers.

It returns the division result when the denominator is not zero.

It returns `None` when the denominator is zero.


```python
def safe_divide(numerator, denominator):
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## Exercise 2

Write a function that converts a list of text values to floats.

Invalid values should be skipped.

The function should return the valid numbers.


```python
def convert_to_floats(raw_values):
    numbers = []

    for raw_value in raw_values:
        try:
            numbers.append(float(raw_value))
        except ValueError:
            continue

    return numbers

raw_values = ["1.5", "2.0", "missing", "3.7"]

print(convert_to_floats(raw_values))
```

## Mini case study

A list contains price values as text.

Some values are missing or invalid.

We want to clean the data and compute a simple report.

The program should not crash because of bad data.


```python
raw_prices = ["100", "101.5", "", "error", "103", "104.2"]

valid_prices = []
invalid_prices = []

for raw_price in raw_prices:
    try:
        valid_prices.append(float(raw_price))
    except ValueError:
        invalid_prices.append(raw_price)

if valid_prices:
    average_price = sum(valid_prices) / len(valid_prices)
    print(f"Average valid price: {average_price:.2f}")
else:
    print("No valid prices were found.")

print("Invalid entries:", invalid_prices)
```

# Introduction to Object Oriented Programming

A program can contain values and functions.

As programs grow these values and functions may become difficult to organize.

Object oriented programming groups related data and behavior together.

A class defines a type of object.


## From dictionaries to objects

A dictionary can store information about an entity.

This is useful for small examples.

For larger programs we may want a more structured representation.


```python
asset = {
    "ticker": "AAPL",
    "name": "Apple Inc.",
    "price": 190.25,
    "shares": 50,
}

market_value = asset["price"] * asset["shares"]

print(f"Market value: {market_value:.2f}")
```

## Classes

A class is a blueprint for creating objects.

An object is an instance of a class.

The class defines what data and behavior the object has.


```python
class Asset:
    pass

asset = Asset()

print(type(asset))
```

## Attributes

Attributes store data inside an object.

They describe the state of the object.

Attributes can be added to an object using dot notation.


```python
class Asset:
    pass

asset = Asset()

asset.ticker = "AAPL"
asset.price = 190.25
asset.shares = 50

print(asset.ticker)
print(asset.price)
print(asset.shares)
```

## The constructor method

The method `__init__` initializes a new object.

It runs automatically when the object is created.

It is used to set the initial attributes.


```python
class Asset:
    def __init__(self, ticker, price, shares):
        self.ticker = ticker
        self.price = price
        self.shares = shares

asset = Asset("AAPL", 190.25, 50)

print(asset.ticker)
print(asset.price)
print(asset.shares)
```

## Methods

A method is a function that belongs to a class.

A method usually works with the data stored in the object.

The parameter `self` refers to the current object.


```python
class Asset:
    def __init__(self, ticker, price, shares):
        self.ticker = ticker
        self.price = price
        self.shares = shares

    def market_value(self):
        return self.price * self.shares

asset = Asset("AAPL", 190.25, 50)

print(f"Market value: {asset.market_value():.2f}")
```

## Creating several objects

A class can be used to create many objects.

Each object has its own attributes.

The method works separately for each object.


```python
asset_1 = Asset("AAPL", 190.25, 50)
asset_2 = Asset("MSFT", 420.10, 20)

print(asset_1.ticker, asset_1.market_value())
print(asset_2.ticker, asset_2.market_value())
```

## A readable representation

The method `__repr__` defines how an object is represented.

This is useful when objects are printed.

A good representation helps with debugging.


```python
class Asset:
    def __init__(self, ticker, price, shares):
        self.ticker = ticker
        self.price = price
        self.shares = shares

    def market_value(self):
        return self.price * self.shares

    def __repr__(self):
        return f"Asset(ticker={self.ticker}, price={self.price}, shares={self.shares})"

asset = Asset("AAPL", 190.25, 50)

print(asset)
```

## Updating an object

Object attributes can be changed.

This is useful when the state of an object changes.

For example a price may change over time.


```python
asset = Asset("AAPL", 190.25, 50)

print(asset.market_value())

asset.price = 195.00

print(asset.market_value())
```

## Adding validation to a class

A class can check whether inputs are valid.

This prevents invalid objects from being created.

The constructor is a natural place for these checks.


```python
class Asset:
    def __init__(self, ticker, price, shares):
        if price <= 0:
            raise ValueError("Price must be positive.")
        if shares < 0:
            raise ValueError("Shares cannot be negative.")

        self.ticker = ticker
        self.price = price
        self.shares = shares

    def market_value(self):
        return self.price * self.shares

try:
    asset = Asset("AAPL", -190.25, 50)
except ValueError as error:
    print(error)
```

## A Portfolio class

A portfolio contains several assets.

This is a good example of composition.

One object can contain other objects.


```python
class Asset:
    def __init__(self, ticker, price, shares):
        if price <= 0:
            raise ValueError("Price must be positive.")
        if shares < 0:
            raise ValueError("Shares cannot be negative.")

        self.ticker = ticker
        self.price = price
        self.shares = shares

    def market_value(self):
        return self.price * self.shares

class Portfolio:
    def __init__(self):
        self.assets = []

    def add_asset(self, asset):
        self.assets.append(asset)

    def total_value(self):
        total = 0

        for asset in self.assets:
            total += asset.market_value()

        return total

portfolio = Portfolio()

portfolio.add_asset(Asset("AAPL", 190.25, 50))
portfolio.add_asset(Asset("MSFT", 420.10, 20))

print(f"Portfolio value: {portfolio.total_value():.2f}")
```

## Portfolio report

A class can also provide reporting methods.

A report method can print the state of the object.

This keeps related behavior close to related data.


```python
class Portfolio:
    def __init__(self):
        self.assets = []

    def add_asset(self, asset):
        self.assets.append(asset)

    def total_value(self):
        return sum(asset.market_value() for asset in self.assets)

    def report(self):
        for asset in self.assets:
            print(f"{asset.ticker}: {asset.market_value():.2f}")
        print(f"Total: {self.total_value():.2f}")

portfolio = Portfolio()

portfolio.add_asset(Asset("AAPL", 190.25, 50))
portfolio.add_asset(Asset("MSFT", 420.10, 20))
portfolio.add_asset(Asset("GOOG", 135.80, 30))

portfolio.report()
```

## A simple BankAccount class

Object oriented programming is not only useful in finance.

It can model many real world entities.

A bank account has a balance and operations that change the balance.


```python
class BankAccount:
    def __init__(self, owner, balance=0):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.balance:
            raise ValueError("Insufficient funds.")
        self.balance -= amount

account = BankAccount("Ana", 1000)

account.deposit(250)
account.withdraw(100)

print(account.owner)
print(account.balance)
```

## Exercise 4

Create a class called `Rectangle`.

The class should store width and height.

Add methods for area and perimeter.


```python
class Rectangle:
    def __init__(self, width, height):
        if width <= 0 or height <= 0:
            raise ValueError("Width and height must be positive.")

        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

rectangle = Rectangle(5, 3)

print(rectangle.area())
print(rectangle.perimeter())
```

## Mini case study

---



We create a small portfolio of positions.

Each position knows how to compute its own value.

The portfolio combines the positions into a total result.


```python
class Position:
    def __init__(self, ticker, shares, buy_price, current_price):
        if shares < 0:
            raise ValueError("Shares cannot be negative.")
        if buy_price <= 0 or current_price <= 0:
            raise ValueError("Prices must be positive.")

        self.ticker = ticker
        self.shares = shares
        self.buy_price = buy_price
        self.current_price = current_price

    def market_value(self):
        return self.shares * self.current_price

    def profit_loss(self):
        return self.shares * (self.current_price - self.buy_price)

class SimplePortfolio:
    def __init__(self, positions):
        self.positions = positions

    def total_value(self):
        return sum(position.market_value() for position in self.positions)

    def total_profit_loss(self):
        return sum(position.profit_loss() for position in self.positions)

    def report(self):
        for position in self.positions:
            print(
                f"{position.ticker}: "
                f"value={position.market_value():.2f}, "
                f"profit_loss={position.profit_loss():.2f}"
            )

        print(f"Total value: {self.total_value():.2f}")
        print(f"Total profit or loss: {self.total_profit_loss():.2f}")

positions = [
    Position("AAPL", 20, 180.00, 190.25),
    Position("MSFT", 10, 390.00, 420.10),
    Position("GOOG", 15, 130.00, 135.80),
]

portfolio = SimplePortfolio(positions)

portfolio.report()
```

##Inheritance

Inheritance allows one class to reuse another class.

The new class inherits the attributes and methods of the existing class.

This reduces code duplication.

It also makes programs easier to extend.

```python
class Asset:
    def __init__(self, ticker, price):
        self.ticker = ticker
        self.price = price

    def display(self):
        print(f"{self.ticker}: {self.price:.2f}")


class Stock(Asset):
    def __init__(self, ticker, price, exchange):
        super().__init__(ticker, price)
        self.exchange = exchange


apple = Stock("AAPL", 190.25, "NASDAQ")

apple.display()

print(f"Exchange: {apple.exchange}")
```

##Extending an Existing Class

A derived class can add new methods.

The inherited methods remain available.

This allows the new class to specialize its behavior.

```python
class Asset:
    def description(self):
        return "Generic financial asset"


class Stock(Asset):
    def description(self):
        return "Equity instrument"


class Bond(Asset):
    def description(self):
        return "Fixed income instrument"


assets = [
    Asset(),
    Stock(),
    Bond()
]

for asset in assets:
    print(asset.description())
```

When to use classes

Classes are useful when data and behavior belong together.

They are useful when a program models entities.

They are less useful for very small scripts.

A simple function is often enough for a simple calculation.
