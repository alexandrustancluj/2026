## Data Structures, Matrices and Comprehensions

This lab covers lists dictionaries sets matrices represented as lists of lists and comprehensions.

## Lists

A list stores several values in one variable.

The values have an order.

Lists are useful when we work with sequences of data.


## Finding the maximum value

This is a classical list algorithm.

We start with the first element.

Then we compare every element with the current maximum.


```python
values = [10, 5, 7, 12, 3]

maximum = values[0]

for value in values:
    if value > maximum:
        maximum = value

print(f"Maximum value: {maximum}")
```

## Finding the minimum value

The minimum algorithm is similar.

Only the comparison changes.

This is a useful pattern for many problems.


```python
values = [10, 5, 7, 12, 3]

minimum = values[0]

for value in values:
    if value < minimum:
        minimum = value

print(f"Minimum value: {minimum}")
```

## Counting values that satisfy a condition

A counter starts from zero.

Each time the condition is true the counter increases.

This pattern appears in many data analysis tasks.


```python
values = [10, -5, 7, -12, 3, 0, -1]

negative_count = 0

for value in values:
    if value < 0:
        negative_count += 1

print(f"Number of negative values: {negative_count}")
```

## Computing the average

The average is the sum divided by the number of values.

We can compute the sum manually.

Then we divide by the length of the list.


```python
values = [10, 5, 7, 12, 3]

total = 0

for value in values:
    total += value

average = total / len(values)

print(f"Average value: {average:.2f}")
```

## Linear search

Linear search checks the elements one by one.

The algorithm stops when the target is found.

This is simple and useful for unsorted data.


```python
values = [10, 5, 7, 12, 3]
target = 7

found = False

for value in values:
    if value == target:
        found = True
        break

print(f"Target found: {found}")
```

## Searching for the first position

Sometimes we need the position of the target.

The function `enumerate()` gives both the index and the value.

This is a common Python pattern.


```python
values = [10, 5, 7, 12, 3]
target = 12

position = None

for index, value in enumerate(values):
    if value == target:
        position = index
        break

print(f"Position: {position}")
```

## Removing duplicates while preserving order

A set can check whether a value has already appeared.

A list keeps the original order.

Together they solve the problem in a clear way.


```python
values = [4, 2, 4, 1, 2, 5, 1, 3]

seen = set()
unique_values = []

for value in values:
    if value not in seen:
        unique_values.append(value)
        seen.add(value)

print(unique_values)
```

## Dictionaries

A dictionary stores key value pairs.

The key is used to access the value.

Dictionaries are useful for frequency tables and labeled information.


```python
student = {
    "name": "Ana",
    "grade": 9,
    "status": "active"
}

print(student["name"])
print(student["grade"])
```

## Frequency table with a dictionary

A frequency table counts how many times each value appears.

The observed value becomes the key.

The count becomes the value.


```python
grades = [10, 8, 9, 10, 7, 8, 10, 9, 6]

frequency = {}

for grade in grades:
    frequency[grade] = frequency.get(grade, 0) + 1

print(frequency)
```

## Dictionary lookup table

A dictionary can be used as a small lookup table.

This avoids long chains of conditions.

The key identifies the value we need.


```python
grade_labels = {
    10: "excellent",
    9: "very good",
    8: "good",
    7: "satisfactory",
    6: "sufficient"
}

grade = 9

print(grade_labels.get(grade, "unknown grade"))
```

## Sets

A set stores unique values.

Sets support mathematical operations.

They are useful for comparing collections.


```python
group_a = {"Ana", "Mihai", "Ioana", "Paul"}
group_b = {"Ioana", "Paul", "Elena", "Radu"}

print("Common:", group_a & group_b)
print("Only in A:", group_a - group_b)
print("Only in B:", group_b - group_a)
```

## Matrix as a list of lists

A matrix can be represented as a list of rows.

Each row is also a list.

This representation is useful before using NumPy.


## Traversing all matrix elements

A nested loop visits all elements.

The outer loop controls rows.

The inner loop controls columns.


```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [8, 7, 9]
]

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

## Row sums

The sum of each row can be computed with a loop.

Each row produces one result.


```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [8, 7, 9]
]

for row_index, row in enumerate(matrix):
    row_sum = sum(row)
    print(f"Row {row_index}: sum = {row_sum}")
```

## Column sums

Column sums require access by column index.

We loop over column positions.

For each column we add the values from all rows.


```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [8, 7, 9]
]

rows = len(matrix)
columns = len(matrix[0])

for column in range(columns):
    column_sum = 0

    for row in range(rows):
        column_sum += matrix[row][column]

    print(f"Column {column}: sum = {column_sum}")
```

## Main diagonal

The main diagonal contains elements whose row index equals the column index.

This algorithm works for square matrices.


```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [8, 7, 9]
]

main_diagonal = []

for index in range(len(matrix)):
    main_diagonal.append(matrix[index][index])

print(main_diagonal)
print(f"Main diagonal sum: {sum(main_diagonal)}")
```

## Secondary diagonal

The secondary diagonal goes from the top right to the bottom left.

For row `i` the column is `n - 1 - i`.


```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [8, 7, 9]
]

n = len(matrix)
secondary_diagonal = []

for i in range(n):
    secondary_diagonal.append(matrix[i][n - 1 - i])

print(secondary_diagonal)
print(f"Secondary diagonal sum: {sum(secondary_diagonal)}")
```

## Transposing a matrix

The transpose of a matrix changes rows into columns.

The element at position `(i, j)` becomes the element at position `(j, i)`.

This is one of the most important classical matrix algorithms.


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

rows = len(matrix)
columns = len(matrix[0])

transpose = []

for column in range(columns):
    new_row = []

    for row in range(rows):
        new_row.append(matrix[row][column])

    transpose.append(new_row)

print("Original matrix:")
for row in matrix:
    print(row)

print("Transposed matrix:")
for row in transpose:
    print(row)
```

## Matrix addition

Two matrices can be added when they have the same dimensions.

Elements in the same position are added together.

The result is a new matrix.


```python
a = [
    [1, 2, 3],
    [4, 5, 6]
]

b = [
    [10, 20, 30],
    [40, 50, 60]
]

result = []

for i in range(len(a)):
    row = []

    for j in range(len(a[0])):
        row.append(a[i][j] + b[i][j])

    result.append(row)

for row in result:
    print(row)
```

## Scalar multiplication

Scalar multiplication multiplies every matrix element by the same value.

The result has the same dimensions as the original matrix.


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

scalar = 3
result = []

for row in matrix:
    new_row = []

    for value in row:
        new_row.append(scalar * value)

    result.append(new_row)

for row in result:
    print(row)
```

## Matrix multiplication

Matrix multiplication combines rows from the first matrix with columns from the second matrix.

The number of columns in the first matrix must match the number of rows in the second matrix.

This implementation uses three nested loops.


```python
a = [
    [1, 2],
    [3, 4]
]

b = [
    [5, 6],
    [7, 8]
]

rows_a = len(a)
columns_a = len(a[0])
columns_b = len(b[0])

result = []

for i in range(rows_a):
    result_row = []

    for j in range(columns_b):
        value = 0

        for k in range(columns_a):
            value += a[i][k] * b[k][j]

        result_row.append(value)

    result.append(result_row)

for row in result:
    print(row)
```

## Comprehensions

A comprehension is a compact way to create a new collection.

It can transform data.

It can also filter data.

A comprehension should be used only when the result is readable.


```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

## Filtering with a comprehension

A condition can be added to the comprehension.

Only values that satisfy the condition are kept.


```python
values = [10, -5, 7, -12, 3, 0, -1]

positive_values = [value for value in values if value > 0]

print(positive_values)
```

## Matrix transpose with comprehension

A matrix transpose can be implemented with a nested comprehension.

This version is shorter than the loop version.

The loop version is usually easier for beginners to understand.


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

transpose = [
    [matrix[row][column] for row in range(len(matrix))]
    for column in range(len(matrix[0]))
]

for row in transpose:
    print(row)
```

## Matrix addition with comprehension

Matrix addition can also be written with nested comprehensions.

This is compact and Pythonic.

It should only be used after the loop version is understood.


```python
a = [
    [1, 2, 3],
    [4, 5, 6]
]

b = [
    [10, 20, 30],
    [40, 50, 60]
]

result = [
    [a[i][j] + b[i][j] for j in range(len(a[0]))]
    for i in range(len(a))
]

for row in result:
    print(row)
```

## Dictionary comprehension

A dictionary comprehension creates a dictionary.

The next example stores the frequency of unique values.

This combines a set and a dictionary comprehension.


```python
grades = [10, 8, 9, 10, 7, 8, 10, 9, 6]

frequency = {grade: grades.count(grade) for grade in set(grades)}

print(frequency)
```

## Set comprehension

A set comprehension creates a set.

Duplicates are removed automatically.

This is useful when we need unique transformed values.


```python
names = ["ana", "Mihai", "ANA", "maria", "Mihai"]

normalized_names = {name.lower() for name in names}

print(normalized_names)
```

Find the largest element in a matrix.

Use a nested loop.

Start from the first element.


```python
matrix = [
    [3, 8, 1],
    [9, 2, 7],
    [4, 6, 5]
]

maximum = matrix[0][0]

for row in matrix:
    for value in row:
        if value > maximum:
            maximum = value

print(maximum)
```

## Practice exercise 3

Check whether a square matrix is symmetric.

A matrix is symmetric when `matrix[i][j] == matrix[j][i]`.

This is a classical algorithm that uses two indexes.


```python
matrix = [
    [1, 2, 3],
    [2, 5, 6],
    [3, 6, 9]
]

is_symmetric = True
n = len(matrix)

for i in range(n):
    for j in range(n):
        if matrix[i][j] != matrix[j][i]:
            is_symmetric = False
            break

print(f"Symmetric matrix: {is_symmetric}")
```
